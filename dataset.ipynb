# Credit Risk Dataset

Loads the [`laotse/credit-risk-dataset`](https://www.kaggle.com/datasets/laotse/credit-risk-dataset) dataset from Kaggle via `kagglehub` and previews it.

**Contents:**
- `fetch_dataset()` — downloads the dataset and loads `credit_risk_dataset.csv` into a pandas `DataFrame`.
- Preview cells — call `fetch_dataset()`, then display the column dtypes and a sample of rows.

```python
import os
import kagglehub
from pandas import DataFrame, pandas as pd

DATASET = "laotse/credit-risk-dataset"

def fetch_dataset() -> DataFrame:
    dataset_dir = kagglehub.dataset_download("laotse/credit-risk-dataset")
    file_path = os.path.join(dataset_dir,"credit_risk_dataset.csv")

    return pd.read_csv(file_path)
```

```python
from IPython.display import display

df = fetch_dataset()

display("\nColumns and data types:\n")
display(df.dtypes)
```

```python
display("\nDataset sample:\n")
display(df.head())
```
